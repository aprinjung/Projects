```python
# 한국은행 api로 회사채3년 데이터 가져오기
url = 'https://ecos.bok.or.kr/api/StatisticSearch/' + apikey \
        + '/json/kr/1/100/817Y002/D/20200101/20250131/010300000'
response = requests.get(url)
result = response.json()
list_total_count=(int)(result['StatisticSearch']['list_total_count'])
list_count=(int)(list_total_count/100) + 1


rows=[]
for i in range(0,list_count):
    start = str(i * 100 + 1)
    end = str((i + 1) * 100)
    
    url = 'https://ecos.bok.or.kr/api/StatisticSearch/' + apikey + '/json/kr/' \
            + start + '/' + end + '/817Y002/D/20200101/20250131/010300000'
    response = requests.get(url)
    result = response.json()
    rows = rows + result['StatisticSearch']['row']
    
company_df=pd.DataFrame(rows)

company_df=company_df[['ITEM_NAME1','TIME','DATA_VALUE']]
company_df['date']=pd.to_datetime((company_df['TIME'].str[:4] + '-' + company_df['TIME'].str[4:6] + '-'+ company_df['TIME'].str[6:8]))
company_df=company_df.astype({'DATA_VALUE':'float'})
company_df=company_df.drop_duplicates()
company_df
```

```python
# 신용스프레드 계산
dfbond=pd.concat([country_df.set_index('date')[['DATA_VALUE']],company_df.set_index('date')[['DATA_VALUE']]], axis=1)
dfbond.columns=['국고채 3년','회사채 3년']
dfbond['신용스프레드']= dfbond['회사채 3년'] - dfbond['국고채 3년']
dfbond
```

```python
# 한국은행 api로 한국은행 기준금리 데이터 가져오기
url = 'https://ecos.bok.or.kr/api/StatisticSearch/' + apikey \
        + '/json/kr/1/100/722Y001/D/20200101/20250131/0101000'
response = requests.get(url)
result = response.json()
list_total_count=(int)(result['StatisticSearch']['list_total_count'])
list_count=(int)(list_total_count/100) + 1


rows=[]
for i in range(0,list_count):
    start = str(i * 100 + 1)
    end = str((i + 1) * 100)
    
    url = 'https://ecos.bok.or.kr/api/StatisticSearch/' + apikey + '/json/kr/' \
            + start + '/' + end + '/722Y001/D/20200101/20250131/0101000'
    response = requests.get(url)
    result = response.json()
    rows = rows + result['StatisticSearch']['row']
    
df=pd.DataFrame(rows)
df
interest_df=pd.DataFrame(rows)

interest_df=interest_df[['ITEM_NAME1','TIME','DATA_VALUE']]
interest_df['date']=pd.to_datetime((interest_df['TIME'].str[:4] + '-' + interest_df['TIME'].str[4:6] + '-'+ interest_df['TIME'].str[6:8]))
interest_df=interest_df.astype({'DATA_VALUE':'float'})
interest_df=interest_df.drop_duplicates()
interest_df
```

```python
# 신용스프레드 df와 인덱스(날짜) 일치하는 행만 가져옴
interest_df = interest_df[interest_df['date'].isin(dfbond.index)]
interest_df
```

```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler

interest = interest_df.DATA_VALUE.tolist()
spread = dfbond.신용스프레드.tolist()

# 데이터 (0 또는 음수 값이 없는 양수 데이터)
interest_data = np.array(interest).reshape(-1, 1)
spread_data = np.array(spread).reshape(-1, 1)

# 로그 변환
log_interest_data = np.log1p(interest_data)  # log(x+1)
log_spread_data = np.log1p(spread_data)  # log(x+1)

# Min-Max Scaling (0과 1 사이로 정규화)
scaler = MinMaxScaler()
normalized_interest_data = scaler.fit_transform(log_interest_data)
normalized_spread_data = scaler.fit_transform(log_spread_data)
```

```python
# 국고채 vs 회사채 그래프 그리기
import plotly.express as px
from plotly.subplots import make_subplots
import plotly.graph_objects as go
import plotly.offline as offline

# 오프라인 모드 설정
offline.init_notebook_mode(connected=True)

fig = go.Figure()
fig.add_trace(
    go.Scatter(x=country_df.date, y=country_df['DATA_VALUE'], name="국채 3년"),
)
fig.add_trace(
    go.Scatter(x=company_df.date, y=company_df['DATA_VALUE'], name="회사채 3년(AA-)"),
)

fig.update_layout(title_text='국채 3년과 회사채 3년(AA-)', title_x=0.5)

# 이미지 파일로 저장
offline.plot(fig, filename='./graph/국채 vs 회사채.png', auto_open=False, image='png')

fig.show()
```

```python
# 신용스프레드 vs 금리 그래프 그리기
import matplotlib.pyplot as plt

# 날짜 데이터 생성 (X축)
dates = interest_df.date

# 그래프 그리기
plt.figure(figsize=(12, 6))  # 그래프 크기 설정 (선택 사항)

plt.plot(dates, normalized_interest_data, label='Normalized Log Interest Rate')
plt.plot(dates, normalized_spread_data, label='Normalized Log Credit Spread')

plt.xlabel('Date')
plt.ylabel('Normalized Value')
plt.title('Normalized Log Interest Rate vs. Normalized Log Credit Spread')
plt.legend()  # 범례 표시
plt.grid(True)  # 격자 추가 (선택 사항)
plt.xticks(rotation=45)  # X축 레이블 회전 (선택 사항)
plt.tight_layout()  # 레이아웃 조정 (선택 사항)
plt.savefig('./graph/신용스프레드 vs 금리.png')
plt.show()
```
